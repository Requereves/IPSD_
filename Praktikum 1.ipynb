2. **Apakah hasil sesuai dugaan awal? :** Sesuai seperti yang saya harapkan yaitu menunjukkan rata2 dan performa mahasiswa
3. **Keterbatasan analisis :** Data hanya 30 mahasiswa dan tidak menjelaskan kondisi setiap mahasiswa serta hubungan antar variabel
4. **Apakah diperlukan data tambahan :** Iya, seperti jenis matkul, nilai ujian, nilai tugas


```python
# Challenge #

def analyze_student(dataframe, student_id):
    try:
        # buat nyari nama Siswa pake student_id
        student = dataframe[dataframe["student_id"] == student_id]
    
        # cek apakah id siswa ada?
        if len(student) == 0:
            print("Siswa tidak ada")
        else: 
            study_hours = student["study_hours"].values[0]
            attendance = student["attendance"].values[0]
            quiz_score = student["quiz_score"].values[0]
        
        # menentukan status siswa
            if study_hours >= 6 and attendance >= 75 and quiz_score >= 70:
                status = "On Track"
                recommendation = "Tahanin belajar dan kehadiran."
            else:
                status = "Needs Review"
                recommendation = "Tingkatkan belajar atau kehadiran."
        
        #output 
        print("Student ID:", student_id)
        print("Jam Belajar:", study_hours, "jam")
        print("Kehadiran:", attendance, "%")
        print("Nilai Kuis:", quiz_score)
        print("Status:", status)
        print("Rekomendasi:", recommendation)
        
    except ValueError as e:
        print(e)
        
analyze_student(df, 1)
```
